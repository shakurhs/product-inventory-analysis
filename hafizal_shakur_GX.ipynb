```python
# import librabries
from great_expectations.data_context import FileDataContext
```

```python
# mendefinisikan context dan membuat directory
context = FileDataContext.create(project_root_dir=r"C:\temp")
```

```python
# membuat data source
datasource_name = 'project-m3-v1'
datasource = context.sources.add_pandas(datasource_name)

# membuat asset
asset_name = 'supermarket_inventory'
path_to_data = r"C:\Users\hafiz\OneDrive\Documents\Hacktiv8_Data_Science_Bootcamp\rmt_40\PHASE_2\p2-ftds040-rmt-m3-shakurhs\P2M3_hafizal_shakur_data_clean.csv"
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

batch_request = asset.build_batch_request()
```

```python
# membuat expectation suite dan validator
expectation_suite_name = 'expectation-supermarket-inventory'
context.add_or_update_expectation_suite(expectation_suite_name)

validator = context.get_validator(
    batch_request=batch_request,
    expectation_suite_name=expectation_suite_name
)

validator.head()
```

```python
# validasi 1: column bersifat unik
validator.expect_column_values_to_be_unique('product_id')
```

```python
# validasi 2: rentang value pada sebuah column
validator.expect_column_values_to_be_between(
    column='stock_quantity', min_value=0, max_value=100
)
```

```python
# validasi 3: nilai value pada sebuah column
validator.expect_column_values_to_be_in_set(
    column='status', 
    value_set=['Active', 'Discontinued', 'Backordered']
    )
```

```python
# validasi 4: tipe data pada column
validator.expect_column_values_to_be_in_type_list(
    column='unit_price',
    type_list=["float"]
)
```

```python
# validasi 5: kesesuaian jumlah column
validator.expect_table_column_count_to_equal(
    value=16
)
```

```python
# validasi 6: kesesuaian nama-nama column
validator.expect_table_columns_to_match_ordered_list(
    column_list=['product_id','product_name','category','supplier_id','supplier_name','stock_quantity','reorder_level','reorder_quantity','unit_price','date_received','last_order_date','expiration_date','warehouse_location','sales_volume','inventory_turnover_rate','status']
)
```

```python
# validasi 7: kesesuaian jumlah row
validator.expect_table_row_count_to_equal(
    value=989
)
```

Output yang dihasilkan dari keseluruhan jumlah ekspektasi, dengan total 7, semuanya menghasilkan jawaban `true`. Artinya, setiap column yang diuji menggunakan Great Expectations sudah memenuhi kriteria yang diinginkan. Proses selanjutnya adalah menyimpan ke dalam expectation suite, membuat checkpoint, dan document.

```python
# menyimpan ke expectation suite
validator.save_expectation_suite(discard_failed_expectations=False)
```

```python
# membuat checkpoint
checkpoint_1 = context.add_or_update_checkpoint(
    name='checkpoint_1',
    validator=validator
)

checkpoint_1_result = checkpoint_1.run()
```
